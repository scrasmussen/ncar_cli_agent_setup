# Codex (ChatGPT) Setup
The following are instructions for setting up [Codex](https://learn.chatgpt.com/docs/codex/cli) on Derecho.

**The free ChatGPT plan includes limited Codex access, link to [price tiers](https://chatgpt.com/#pricing)**

# 1. Install locally on Derecho
The Codex install.sh script downloads, installs, and cleans-up the Codex installation files.

After installing, you'll have:
```bash
~/.local/bin/codex          # launcher/symlink (what goes on your PATH)
~/.codex/packages/          # versioned Codex binaries; standalone updates land here
~/.codex/                   # settings, credentials, session history, logs
```

To install, run the following command:

```bash
$ curl -fsSL https://chatgpt.com/codex/install.sh | sh
```

# 2. Add it to your PATH if ~/.local/bin isn't there already:

``` bash
$ echo 'export PATH="$HOME/.local/bin:$PATH"' >> ~/.bashrc
$ source ~/.bashrc
```

# 3. Check it works

``` bash
$ codex --version
codex-cli 0.160.1
```

# 4. Log in
On a headless server like Derecho, use Codex's device-code login.
Codex gives you a URL and a one-time code.

```bash
$ cd /path/to/your/project
$ codex login --device-auth
```

After entering the previous command:
- open the URL in your local browser
- sign in with your ChatGPT account
- enter the code shown in the terminal.

## Quick tips
- Start in your project directory: `cd my-project && codex`
- Add an `AGENTS.md` at the root of your repository with build commands, testing instructions, coding conventions, and project-specific rules
- Ask Codex to explain unfamiliar code before asking it to modify it
- Give Codex a specific goal and tell it how to verify the result, such as which tests or build commands to run
- Review `git diff` before accepting changes, and commit often so you can easily roll back
