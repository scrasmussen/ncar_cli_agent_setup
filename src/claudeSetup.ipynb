# Claude Setup
The following are instructions for setting up [Claude Code](https://claude.com/product/claude-code) on Derecho.

**The free Claude.ai plan does not include Claude Code access, so you need a Pro/Max/Team/Enterprise plan or an API key.**

# 1. Install locally on Derecho
The Claude install.sh script downloads, installs, and cleans-up the Claude installation files.

After installing, you'll have:
```bash
~/.local/bin/claude        # launcher (what goes on your PATH)
~/.local/share/claude/     # versioned binaries; auto-updates land here
~/.claude/                 # settings, credentials, session history
```

# 4. Log in
On a headless server like Derecho, Claudes gives you a URL to open in your local browser. 
You sign in there and paste the code back into the terminal. 
You can also log in with an Anthropic API key by setting ANTHROPIC_API_KEY in your environment.

## Quick tips
- Start in your project directory: `cd my-project && claude`
- Run `/init` to generate a `CLAUDE.md` that describes your project to Claude
- Ask Claude to explain code before you ask it to change code
- Review changes before accepting them, and commit often so you can roll back
